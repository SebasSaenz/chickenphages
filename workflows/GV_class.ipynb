## Split fasta file

```python
outdir = 'Fnas'

!mkdir -p {outdir}

with open('/home/ty/Big_data/ChickenPhages/reps_viral_100K.fna', 'r') as fasta:
    conts = fasta.read().split('\n>')
for cont in conts:
    name = cont.split('\n')[0].split('|')[0].strip('>')
    with open(f'{outdir}/{name}.fna', 'w') as fna:
        fna.write(f">{cont.strip('>')}")
    
```
